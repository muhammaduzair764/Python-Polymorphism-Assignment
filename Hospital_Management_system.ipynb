```python
# Hospital Management System:

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
class Doctor(Person):
    def __init__(self, name, age, specialization, fee):
        super().__init__(name, age)
        self.specialization = specialization
        self.fee = fee
    def display_info(self):
        print("Doctor Name:", self.name)
        print("Age:", self.age)
        print("Specialization:", self.specialization)
        print("Consultation Fee:", self.fee)
class Nurse(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department
class Patient(Person):
    def __init__(self, name, age, disease):
        super().__init__(name, age)
        self.disease = disease
        self.appointment = None
# Doctors list
doctors = [
    Doctor("Ali", 40, "Cardiologist", 3000),
    Doctor("Ahmed", 35, "Dermatologist", 2000),
    Doctor("Usman", 45, "Neurologist", 5000)
]


# Display available doctors
print("\n--- Available Doctors ---")

for i, doctor in enumerate(doctors, start=1):
    print(i, "-", doctor.name, "|", doctor.specialization,
          "| Fee:", doctor.fee)


# Get patient information from user
print("\n--- Patient Information ---")

name = input("Enter Patient Name: ")
age = int(input("Enter Patient Age: "))
disease = input("Enter Disease: ")

p1 = Patient(name, age, disease)


# Select doctor
choice = int(input("\nSelect Doctor Number: "))

if choice >= 1 and choice <= len(doctors):
    selected_doctor = doctors[choice - 1]

    # Book appointment
    p1.appointment = selected_doctor

    print("\nAppointment Booked Successfully")
    print("Patient:", p1.name)
    print("Doctor:", selected_doctor.name)

    # Generate bill
    print("\n--- Hospital Bill ---")
    print("Patient Name:", p1.name)
    print("Doctor Name:", selected_doctor.name)
    print("Consultation Fee:", selected_doctor.fee)
    print("Total Bill:", selected_doctor.fee)

    # Cancel appointment
    cancel = input("\nDo you want to cancel appointment? (yes/no): ")

    if cancel.lower() == "yes":
        p1.appointment = None
        print("Appointment Cancelled Successfully")
    else:
        print("Appointment is Confirmed")

else:
    print("Invalid Doctor Selection")
```

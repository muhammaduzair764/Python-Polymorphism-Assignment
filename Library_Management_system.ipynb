```python
# Library Management System

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

    def access_book(self):
        print("Books can be accessed")


class Ebook(Book):
    def access_book(self):
        print("Ebook can be read online")


class Printbook(Book):
    def access_book(self):
        print("Printed book can be read physically")


class AudioBook(Book):
    def access_book(self):
        print("AudioBook can be listened to")


# Get books from user
books = []
print("<--- Enter Books Details --->")
for i in range(3):
    print("\nBook", i + 1)
    title = input("Enter book title: ")
    author = input("Enter book author: ")
    print("1. Ebooks")
    print("2. Printbooks")
    print("3. Audiobooks")
    choice = int(input("Select book type: "))
    if choice == 1:
        book = Ebook(title, author)

    elif choice == 2:
        book = Printbook(title, author)

    elif choice == 3:
        book = AudioBook(title, author)

    else:
        print("Invalid Book Type")
        continue

    books.append(book)


# Search Book
print("\n--- Search Book ---")

search = input("Enter book title to search: ")

found = False

for book in books:

    if book.title.lower() == search.lower():

        found = True

        print("\nBook Found!")
        print("Title:", book.title)
        print("Author:", book.author)

        if book.available:
            print("Status: Available")

            issue = input("Do you want to issue this book? (yes/no): ")

            if issue.lower() == "yes":
                book.available = False
                print("Book issued successfully")

                # Access book
                book.access_book()

        else:
            print("Status: Already Issued")

        break


if not found:
    print("Book not found")


# Return Book
print("\n--- Return Book ---")

return_book = input("Enter book title to return: ")

for book in books:

    if book.title.lower() == return_book.lower():

        if not book.available:
            book.available = True
            print("Book returned successfully")
        else:
            print("This book was not issued")

        break

else:
    print("Book not found")

```

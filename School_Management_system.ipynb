```python
# School Management System
# Inheritance + Method Overriding

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def calculate_result(self):
        total = sum(self.marks)
        percentage = total / 500 * 100

        print("\n--- Student Result ---")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Total Marks:", total)
        print("Percentage:", percentage)

        if percentage >= 80:
            print("Grade: A+")
        elif percentage >= 70:
            print("Grade: A")
        elif percentage >= 60:
            print("Grade: B")
        elif percentage >= 50:
            print("Grade: C")
        else:
            print("Grade: Fail")


class ScholarshipStudent(Student):
    def calculate_result(self):
        total = sum(self.marks)
        percentage = total / 500 * 100

        print("\n--- Scholarship Student Result ---")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Total Marks:", total)
        print("Percentage:", percentage)

        if percentage >= 90:
            print("Grade: A+")
            print("Scholarship: Eligible")
        elif percentage >= 80:
            print("Grade: A")
            print("Scholarship: Eligible")
        elif percentage >= 70:
            print("Grade: B")
            print("Scholarship: Not Eligible")
        else:
            print("Grade: Fail")
            print("Scholarship: Not Eligible")


class Teacher(Person):
    def teaching(self):
        print(self.name, "is teaching")


# Get student information
name = input("Enter Student Name: ")
age = int(input("Enter Student Age: "))

# Get marks of 5 subjects
marks = []

print("\nEnter marks of 5 subjects:")

for i in range(5):
    mark = float(input("Enter marks for Subject " + str(i + 1) + ": "))
    marks.append(mark)


# Select student type
print("\n1. Regular Student")
print("2. Scholarship Student")

choice = int(input("Select Student Type: "))


if choice == 1:
    student = Student(name, age, marks)

elif choice == 2:
    student = ScholarshipStudent(name, age, marks)

else:
    print("Invalid Student Type")
    exit()


# Display result
student.calculate_result()
```

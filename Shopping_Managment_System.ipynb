```python
# Shopping Management System
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def display(self):
        print("Product:", self.name)
        print("Price:", self.price)
class Cart:
    def __init__(self):
        self.products = []
    def add_to_cart(self, product):
        self.products.append(product)
        print(product.name, "added to cart.")
    def remove_from_cart(self, name):
        for product in self.products:
            if product.name.lower() == name.lower():
                self.products.remove(product)
                print(name, "removed from cart.")
                return
        print("Product not found in cart.")
    def show_cart(self):
        print("\n--- Shopping Cart ---")
        if len(self.products) == 0:
            print("Cart is empty.")
            return
        total = 0
        for product in self.products:
            print(product.name, "-", product.price)
            total += product.price
        print("Total Bill:", total)
# Get product 1 data from user
print("Enter Laptop Details:")
name = input("Enter product name: ")
price = int(input("Enter product price: "))
laptop = Product(name, price)
# Get product 2 data from user
print("\nEnter Mobile Details:")
name = input("Enter product name: ")
price = int(input("Enter product price: "))
mobile = Product(name, price)
# Get product 3 data from user
print("\nEnter Shirt Details:")
name = input("Enter product name: ")
price = int(input("Enter product price: "))
shirt = Product(name, price)
# Create Cart
cart = Cart()
# Add products to cart
cart.add_to_cart(laptop)
cart.add_to_cart(shirt)
cart.add_to_cart(mobile)
# Show cart
cart.show_cart()
# Remove product
remove_name = input("\nEnter product name to remove: ")
cart.remove_from_cart(remove_name)
# Show updated cart
cart.show_cart()
```

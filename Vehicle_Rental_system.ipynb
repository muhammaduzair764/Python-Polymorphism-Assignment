```python
# Vehicle Rental Management System

class Vehicle:
    def __init__(self, name, rent_per_day, days):
        self.name = name
        self.rent_per_day = rent_per_day
        self.days = days

    def calculate_rent(self):
        return self.rent_per_day * self.days

    def display(self):
        print("Vehicle:", self.name)
        print("Rent per day: Rs.", self.rent_per_day)
        print("Days:", self.days)


# Child Class 1
class Car(Vehicle):
    def calculate_rent(self):
        # Car has 10% extra charges
        rent = self.rent_per_day * self.days
        return rent * 1.10


# Child Class 2
class Bike(Vehicle):
    def calculate_rent(self):
        # Bike has 5% discount
        rent = self.rent_per_day * self.days
        return rent * 0.95


# Child Class 3
class Bus(Vehicle):
    def calculate_rent(self):
        # Bus has 15% extra charges
        rent = self.rent_per_day * self.days
        return rent * 1.15


# Rental Class
class Rental:
    def __init__(self):
        self.vehicles = []

    def rent_vehicle(self, vehicle):
        self.vehicles.append(vehicle)
        print(vehicle.name, "rented successfully.")

    def return_vehicle(self, name):
        for vehicle in self.vehicles:
            if vehicle.name.lower() == name.lower():
                self.vehicles.remove(vehicle)
                print(vehicle.name, "returned successfully.")
                return

        print("Vehicle not found.")

    def display_rented_vehicles(self):
        print("\n----- Rental Details -----")

        if len(self.vehicles) == 0:
            print("No vehicles rented.")
            return

        total = 0

        for vehicle in self.vehicles:
            rent = vehicle.calculate_rent()
            print("\nVehicle:", vehicle.name)
            print("Days:", vehicle.days)
            print("Rent: Rs.", rent)

            total += rent

        print("--------------------------")
        print("Total Rent: Rs.", total)


# -------------------------
# Enter Car Details
# -------------------------

print("\n<----- Enter Car Details ----->")

name = input("Enter Car Name: ")
rent = int(input("Enter rent per day: "))
days = int(input("Enter days: "))

car = Car(name, rent, days)


# -------------------------
# Enter Bike Details
# -------------------------

print("\n<----- Enter Bike Details ----->")

name = input("Enter Bike Name: ")
rent = int(input("Enter rent per day: "))
days = int(input("Enter days: "))

bike = Bike(name, rent, days)


# -------------------------
# Enter Bus Details
# -------------------------

print("\n<----- Enter Bus Details ----->")

name = input("Enter Bus Name: ")
rent = int(input("Enter rent per day: "))
days = int(input("Enter days: "))

bus = Bus(name, rent, days)


# Create Rental Object
rental = Rental()


# Rent Vehicles
rental.rent_vehicle(car)
rental.rent_vehicle(bike)
rental.rent_vehicle(bus)


# Display Rental Details
rental.display_rented_vehicles()


# Return a Vehicle
return_name = input("\nEnter vehicle name to return: ")

rental.return_vehicle(return_name)


# Display Updated Details
rental.display_rented_vehicles()
```
